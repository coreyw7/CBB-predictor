## Exploring data in DuckDB

```python
con.sql("SELECT AVG(Win) FROM team_games WHERE Loc = 'H'").df()
```

```python
con.sql("SELECT COUNT(DISTINCT TeamID) FROM team_games").df()
```

```python
con.sql("SELECT Season, avg(Possessions) AS avg_possessions FROM team_games GROUP BY Season ORDER BY Season").df()
```

```python
con.sql("""
    SELECT t.TeamName, g.Season, AVG(g.Win) AS win_pct, COUNT(*) AS games, AVG(g.Score - g.OppScore) as avg_margin
    FROM team_games AS g
    JOIN teams AS t ON g.TeamID = t.TeamID
    GROUP BY t.TeamName, g.Season
    ORDER BY win_pct DESC, avg_margin DESC
    LIMIT 5
""").df()
```

```python
con.sql("""
    SELECT t.TeamName, g.Season, AVG(g.Win) AS win_pct, COUNT(*) AS games, AVG(g.Score - g.OppScore) as avg_margin
    FROM team_games AS g
    JOIN teams AS t ON g.TeamID = t.TeamID
    WHERE t.TeamName = 'Arkansas'
    GROUP BY t.TeamName, g.Season
    HAVING COUNT(*) >= 20
    ORDER BY win_pct DESC
    LIMIT 5
""").df()
```

```python
con.close()
```
